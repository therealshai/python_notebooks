```python
def print_hi(name):

    print("This is inside function:"+name)
    print(type(name))
```

```python
def method_slice(start:int, stop:int, step:int, name:str):
    print("This function slices your name starting from "+ str(start)+" till"+ str(stop)+ "by stepping:"+str(step))
    print("your name now is:"+ name[start:stop:step])
```

```python

def method_split(name):
    print("splitting your name /acc to whitespaces")
    print(name.split()) # split based on whitespaces
    print(name.split('i'))
```

```python
def method_format():
    print("Using .format() and f string")

    print("Syed Shaista".format("Tabassum"))
    print("This is a {} example".format("quick"))
    print("this is {1} {0} exmaple of {2} strings".format("quick","another","format"))
    print("this is :{a} {b} {a}".format(a="a",b="b",c="c"))
```

```python
print_hi('Shai')
method_slice(1, 3, 1, 'shai')
method_split("syed shaista")
method_format()
```

- supports indexing and slicing
- it can be nested
- can be mutated

```python
list1=[1,2,3,4,5]
list2=[0,1,2,3,4,5]
list1+list2
```

```python
print(type(list1))
```

```python
list1[0]="ZERO"
print(list1)
```

```python
list1.append("SIX")
print(list1)
```

```python
list1.pop() #remove last element
```

```python
try:
    list1.sort() #sorted in mathematical order
except :
    print("cant sort int and str list")
```

### Dictionaries
- unordered mappings + key:value pairs
- {'key1':value1, 'key2':'value2'}
- cant be sorted

```python
dict1= {
    "abc": "alphabets",
    "123":"numbers",
    "0123": "whole numbers",
    "negative numbers": [-1,0,1],  #value as list
    "positive numbers": {"zero": [0]}  #nested dict
}
```

```python
dict1["123"]
```

```python
type(dict1)
```

```python
print(dict1["negative numbers"])#converted into list
type(dict1["negative numbers"])
```

```python
dict1["positive numbers"]["zero"]
```

```python
dict1["positive numbers"]["zero"] =[0,1,2]
```

```python
print(dict1)
```

```python
dict1.keys() #get all keys
```

```python
dict1.values() #values
```

```python
dict1.items() #pairs of tuples
```

### Tuples
- immutable => basically cant be reassigned
- () -> similar to lists
- can have any object type inside it

```python
t1=(1,2,3)
```

```python
t1
```

```python
len(t1)
```

```python
try:
    t1[0]=0
except:
    print("tuple values cant be reassigned = immutable")
```

```python
type(t1)
```

```python
t1.count(1) #no of times 1 appeared in that tuple
```

```python
t1.index(1) #index of element "1"
```
