```python
import random

def generate_comp_response():
    indexes_for_comp = []
    index = 0

    for row in game:
        for cell in row:
            if cell == 0:
                indexes_for_comp.append(index)
            index += 1

    return random.choice(indexes_for_comp)
```

```python
game= [[0,0,0],
        [0,0,0],
            [0,0,0]]
```

```python
def game_logic():
    """
    Checks the current board state.

    Returns:
        'X' if X wins
        'O' if O wins
        'DRAW' if board is full and no winner
        None if game should continue
    """
    # all possible winning lines (8 total)
    lines = [
        # rows
        [game[0][0], game[0][1], game[0][2]],
        [game[1][0], game[1][1], game[1][2]],
        [game[2][0], game[2][1], game[2][2]],

        # columns
        [game[0][0], game[1][0], game[2][0]],
        [game[0][1], game[1][1], game[2][1]],
        [game[0][2], game[1][2], game[2][2]],

        # diagonals
        [game[0][0], game[1][1], game[2][2]],
        [game[0][2], game[1][1], game[2][0]]
    ]
    print(lines[0])
    # check for winner
    for line in lines:
        if line[0] != 0 and line[0] == line[1] == line[2]:
            print(line[0])
            return line[0]   # 'X' or 'O'

    # check for draw (no empty cells)
    for row in game:
        if 0 in row:
            return None     # game still ongoing

    return "DRAW"
game_logic()
```

```python
def update_matrix(game,index, value,user = True):
    if user == True:
        value='X'
    else:
        value='O'
    if index ==0:
        game[0][0]= value
    elif index ==1:
         game[0][1]= value
    elif index ==2:
        game[0][2]= value
    elif index ==3:
        game[1][0]= value
    elif index ==4:
        game[1][1]= value
    elif index ==5:
        game[1][2]= value
    elif index ==6:
        game[2][0]= value
    elif index ==7:
        game[2][1]= value
    elif index ==8:
        game[2][2]= value


```

```python
def display_game_board(game = game):
    for i in game:
        print(f"\t {i}\n")
display_game_board()
```
