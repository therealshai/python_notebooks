### For loops
many objects in python are iterables => it goes through items that are in a sequence or any other iterable item.

```python
# for loops
for i in range(1,10,2):
    print(i)
```

```python
#iterable => ex: LIST
L1= [1,2,3,4,5]
length =len(L1)
for i in L1:
    if i> length-1:
        print(str(i)+ "\tend of list ")
        break
    print(i)
```

```python
sum=0
for item in L1:
    sum +=item
print(sum)
```

```python
for _ in "shai":  # _ is a common syntax used when var isnt required to perform an operation inside the loop
    print(_)
```

```python
# tuple unpacking
tup1 =((1,2),(3,4))
tup_list1= list(tup1)
for i in tup_list1:
    print(i)

for (i,j) in tup_list1:
    print(i)
    print(j) #unpacked items
print("choose 2nd var in the tuple ")
for (_,j)in tup_list1:
    print(j)
```

```python
d={"k1":1,"k2":2,"k3":3}
# iter over keys

for keys in d:
    print(keys)
```

```python
for val in d.values():
    print(val)
```

```python
for (_,n)in d.items():
    print(_,n)
```

### Break , Continue , Pass

- break: Breaks out of the current closest enclosing loop.
- continue: Goes to the top of the closest enclosing loop.
- pass: Does nothing at all.


```python
x=[1,2,3]
for i in x:
    if i==2:
        pass
    else:
        print(i)
```

```python
for letter in 'shai':
    if letter =="a":
        break  #exits the loop
    else:
        print(letter)
```

```python
for letter in 'shai':
    if letter =="s":
        continue #skips the iteration
    else:
        print(letter)
```
