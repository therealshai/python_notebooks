```python
class circle():
    PI = 3.14 #class object attribute
    def __init__(self,r):
        self.radius=r

    def cal_area(self):
        return self.PI*self.radius**2  #class object attributes can be called : self.PI or circle.PI => use class name to ref it

    def cal_circumference(self):
        return 2*circle.PI*self.radius

    def get_info_about_circle(self):
        area= self.cal_area()
        cirum= self.cal_circumference()

        print(f"The object is of type {type(self)} and is create at {self}")
        print(f"The area of circle is {area} ")
        print(f"The perimeter of circle is {cirum} ")
```

```python
c1= circle(5)
```

# Inheritance and Polymorphism

```python
# inheritance -> parent and child class hierachy
class Animal():
    def __init__(self):
        print("Animal is created")

    def about_me(self):
        print("I am an animal")

class Dog (Animal):  # dog class inherits from animal class
    def __init__(self):
        Animal.__init__(self)  # here when dog class instance is created -> animal class instance is created first
        print("I am a dog")

    def my_action(self):
        print("I like to bark!")

```

```python
dog1= Dog()
```

```python
type(dog1)
```

```python
dog1.my_action() # added a new method
```

```python
#ploymorphism
class Cat():
    def __init__(self):
        print("Cat is created")
    def my_action(self):
        print("I like to meow!")
```

```python
niko = Cat()
niko.my_action()
# same method name , different objects
shiro= Dog()
shiro.my_action()
```

# Special / Magic / Dundder methods

```python
class Book():
    def __init__(self, title, author, pages:int):
        self.title = title
        self.author = author
        self.pages = pages

    # for a string call ( if any method asks for a string rep of the book object)
    def __str__(self):
        return f"This is a Book: {self.title} ,By {self.author} , [{self.pages}]"

    #for length method
    def __len__(self): #len of book = no of pages
        return self.pages

    def __del__(self):
        print("Book is deleted!!")
```

```python
Book1= Book("Python", "Shai", 200)
```

```python
print(Book1) # print function takes the string representation of Book1 => calls the __str__ method
```

```python
len(Book1)
```

```python
del Book1
```

```python
class Line(object):

    def __init__(self,coordinates): #(x,y)
        self.coor1, self.coor2 = coordinates

    def distance(self):
        x1,y1 = self.coor1
        x2,y2 = self.coor2
        return ((x2-x1)**2 + (y2-y1)**2)**0.5

    def slope(self):
        x1,y1 = self.coor1
        x2,y2 = self.coor2
        return (y2-y1)/(x2-x1)

# Fill in the Line class methods to accept coordinates as a pair of tuples and return the slope and distance of the line.
```

```python
list_of_coordinates= ((3,2),(8,10))
li = Line(list_of_coordinates)
```

```python
li.distance()
```

```python
li.slope()
```

# Object Oriented Programming Challenge
For this challenge, create a bank account class that has two attributes:

- owner
- balance
and two methods:
 - deposit
  - withdraw
As an added requirement, withdrawals may not exceed the available balance.

Instantiate your class, make several deposits and withdrawals, and test to make sure the account can't be overdrawn.

```python
class Account:
    def __init__(self,owner,balance):
        self.acc_holder_name = owner
        self.acc_balance = balance

    def __str__(self):
        return f" This account belongs {self.acc_holder_name}"

    def deposit(self,amount):
        self.acc_balance += amount
        return self.acc_balance

    def withdraw(self,amount):
        if amount > self.acc_balance:
            return f"Insufficient funds to withdraw {amount}"
        else:
            self.acc_balance -= amount
            return self.acc_balance

```

```python

acct1 = Account('Jose',100)
# 2. Print the object
```
