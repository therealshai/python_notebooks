# Errors and Exception Handling

- try: code
- except : block of code that will execute if there is any error in the try block
- finally: always runs regardless of the try-block results


```python
def add(a,b):
    return a+b
```

```python
add(10,20)
```

```python
a= 10
b= input("Enter a number")
try:
    print(add(a,b)) # has error
except TypeError:
    print("Cant use this function for  different datatypes")
```

```python
try:
    result=10+10
except TypeError:  # this block runs if there is an error in the try block
    print("Cant use this function for  different datatypes")
else: #this block runs if there is no exception => except block is false
    print("No Exceptions")
    print(result)
finally:
    print("This block always runs")
```

```python
try:
    f=open("test.txt","r")
    f.write("This is a test file")
except FileNotFoundError:
    print("File not exist")
except IOError:
    print("IO Error occured")
finally:
    f.close() # always close the file
```

```python
try:
    for i in ['a','b','c']:
        print(i**2)
except TypeError:
        print("i is not a int, cant use '**' operator")
```

```python
try:
    x = 5
    y = 0
    z = x/y
except ZeroDivisionError:
    print("Division by zero")
finally:
    print("All Done")
```

```python
def ask():
    while True:
        try:
            i = int(input("Enter a number"))
            var = i ** 2
        except ValueError:
            print("An error occurred! Please try again!")
        else:
            print("Thank you, your number squared is: {}".format(var))
            break

```

```python
ask()
```
